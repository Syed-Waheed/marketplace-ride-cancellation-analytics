This is one of the most important analytical stages of **Project 1**.

---

## Step 7 Roadmap

We will proceed through the statistical analysis in the following order:

| Section | Analysis | Status |
|---|---|---|
| **7A** | VTAT ↔ Customer Cancellation — Statistical Significance | 🔵 **NOW** |
| **7B** | VTAT ↔ Cancellation — Effect Size | ⏳ Next |

| **7D** | Multivariable Logistic Regression | ⏳ Upcoming |
| **7E** | Odds Ratios + Confidence Intervals | ⏳ Upcoming |

| **7G** | Final Hypothesis Verdicts | ⏳ Final |

> **Important:** We are **not doing everything at once**.  
> We will complete each stage, interpret the result, and then move to the next one.

# 7A — Does VTAT Actually Relate to Customer Cancellation?

Our descriptive analysis showed a substantial difference in cancellation rates across VTAT groups:

| VTAT Group | Customer Cancellation Rate |
|---|---:|
| **Q1 — Shortest VTAT** | **0.74%** |
| **Q2** | **5.92%** |
| **Q3** | **5.99%** |
| **Q4 — Longest VTAT** | **17.73%** |

At first glance, this appears to be a large difference.

However, a descriptive comparison alone does **not** tell us whether the observed relationship is statistically supported.

So our next question is:

> **Is there a statistically significant association between VTAT group and customer cancellation?**

---

## Why Use a Chi-Square Test?

For this analysis, we have two categorical variables:

**Variable 1 — VTAT Group**

- Q1
- Q2
- Q3
- Q4

**Variable 2 — Customer Cancellation**

- Yes
- No

Because we are testing whether these two categorical variables are associated, we use a:

### **Chi-Square Test of Independence**

The test evaluates whether the distribution of customer cancellations differs across the VTAT groups.

---

## Hypotheses

### **Null Hypothesis (H₀)**

> **VTAT group and customer cancellation are independent.**

In other words, differences in cancellation rates between VTAT groups could be explained by random variation.

### **Alternative Hypothesis (H₁)**

> **VTAT group and customer cancellation are associated.**

In other words, the distribution of customer cancellations differs across VTAT groups.

---

## Significance Level

We will use:

\[
\alpha = 0.05
\]

Therefore:

- **p < 0.05** → Reject H₀
- **p ≥ 0.05** → Fail to reject H₀

> **Important:** A statistically significant result tells us that an association is unlikely to be explained by sampling variation alone. It does **not**, by itself, establish causation.

---

## What We Will Calculate

Our statistical test will provide:

1. **Chi-square statistic (χ²)**
2. **Degrees of freedom (df)**
3. **p-value**

We will then interpret these results in the context of our research question:

> **Is VTAT statistically associated with customer cancellation?**

---

### Next Step

We will construct a **VTAT Group × Customer Cancellation contingency table** and perform the Chi-Square Test of Independence.

````python

from scipy.stats import chi2_contingency

# ---------------------------------------------------
# STEP 7A — Chi-square test
# ---------------------------------------------------

# Use bookings where VTAT is available
stats_df = df[df["Avg_VTAT"].notna()].copy()

# Create VTAT quartiles
stats_df["vtat_quartile"] = pd.qcut(
    stats_df["Avg_VTAT"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"],
    duplicates="drop"
)

# Create contingency table
contingency_table = pd.crosstab(
    stats_df["vtat_quartile"],
    stats_df["is_customer_cancelled"]
)

print("Contingency Table:")
print(contingency_table)

# Chi-square test
chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    contingency_table
)

print("\nChi-square statistic:", chi2)
print("Degrees of freedom:", degrees_of_freedom)
print("p-value:", p_value)

print("\nExpected counts:")
print(expected)
````

## What Are We Looking For?

The key output from the Chi-Square test is the:

### **p-value**

We will evaluate the p-value using our chosen significance level:

\[
\alpha = 0.05
\]

---

## How Do We Interpret the p-value?

### 🟢 If **p < 0.05**

We **reject the null hypothesis (H₀)**.

This means:

> **There is statistically significant evidence that VTAT group and customer cancellation are associated.**

In other words, the differences in cancellation rates across VTAT groups are unlikely to be explained by random variation alone.

---

### 🟡 If **p ≥ 0.05**

We **fail to reject the null hypothesis (H₀)**.

This means:

> **There is not enough statistical evidence to conclude that VTAT group and customer cancellation are associated at the 5% significance level.**

Importantly, this does **not** prove that there is no relationship. It simply means that the evidence is not strong enough to reject the null hypothesis under our chosen significance threshold.

---

# ⚠️ But Don't Stop at the p-value

This is particularly important for this project — and a useful point to remember in interviews.

Our dataset contains approximately **150,000 observations**.

With a large sample size, even relatively small differences can produce statistically significant p-values.

Therefore, our conclusion will **not** simply be:

> ❌ **“p < 0.05, therefore VTAT is important.”**

Instead, we will ask three separate questions:

### 1️⃣ Is the relationship statistically significant?

**Chi-Square Test**

> Is there evidence of an association?

### 2️⃣ How large is the relationship?

**Effect Size**

> Is the association practically meaningful, or merely statistically detectable?

### 3️⃣ Does the relationship remain after controlling for other factors?

**Multivariable Logistic Regression**

> Does VTAT remain associated with customer cancellation when other relevant variables are considered?

This progression takes the analysis beyond a basic statistical test and toward a more robust analytical conclusion.

---

# ⚠️ Association ≠ Causation

One of the most important principles throughout this project is:

> **A statistical association does not establish causation.**

If our Chi-Square test is statistically significant, our conclusion should be:

> **“VTAT group and customer cancellation are statistically associated.”**

We should **not** conclude:

> **“VTAT causes customers to cancel.”**

There may be other factors influencing both VTAT and cancellation.

For example, factors such as:

- vehicle type
- pickup location
- time of day
- customer type
- demand conditions
- other operational factors

may also contribute to the observed relationship.

That is why we will later use **multivariable logistic regression** to examine whether the VTAT relationship remains after controlling for other relevant variables.

---

### 🎯 Analytical Principle

> **Statistical significance tells us whether we have evidence of an association.**
>
> **Effect size tells us how substantial that association is.**
>
> **Multivariable analysis helps us determine whether the relationship remains after accounting for other factors.**
>
> **None of these, by themselves, prove causation.**

# Step 7B — Effect Size: How Strong Is the VTAT Relationship? 📏

Our **7A result** gave us strong statistical evidence of an association between VTAT group and customer cancellation.

The Chi-Square test produced:

> **p-value ≈ 0.0**

### ⚠️ Important clarification

A p-value displayed as `0.0` by Python does **not** necessarily mean that the true p-value is mathematically zero.

It means the p-value is **extremely small** and has been rounded or displayed as `0.0` at the available numerical precision.

Therefore, the appropriate interpretation is:

> **The observed association is statistically significant at the 5% significance level.**

---

# But Statistical Significance Isn't the Whole Story

The Chi-Square test answers:

> **“Is there evidence that VTAT group and customer cancellation are associated?”**

But it does **not** tell us:

> **“How strong is that association?”**

This distinction is particularly important because our dataset contains approximately **150,000 observations**.

With a large sample size, even a relatively small association can produce a very small p-value.

Therefore, we now move from:

### **Statistical significance**

to:

### **Practical significance**

---

# Cramér's V

To measure the strength of association between **VTAT group** and **customer cancellation**, we will use:

## **Cramér's V**

Cramér's V is an effect-size measure for categorical variables.

Its value ranges from:

\[
0 \leq V \leq 1
\]

where:

- **V ≈ 0** → little or no association
- **Larger V** → stronger association
- **V = 1** → perfect association

For our **4 × 2 contingency table**:

- VTAT group → 4 categories
- Customer cancellation → 2 categories

Cramér's V is an appropriate measure of association strength.

---

## Why Are We Calculating It?

Our analysis now has two separate questions:

### Question 1 — Is the relationship statistically significant?

Answered by:

> **Chi-Square Test → p-value**

### Question 2 — How strong is the relationship?

Answered by:

> **Cramér's V → effect size**

This distinction is important.

A result can be:

> **Statistically significant but practically weak**

or:

> **Statistically significant and meaningfully strong**

The p-value alone cannot tell us which situation we are dealing with.

---

# Our Analytical Sequence

We are therefore building the analysis step by step:

| Question | Statistical Tool | What It Tells Us |
|---|---|---|
| Is there an association? | **Chi-Square Test** | Statistical significance |
| How strong is the association? | **Cramér's V** | Effect size |
| Does the relationship remain after controlling for other factors? | **Logistic Regression** | Adjusted association |

---

### 🎯 Key Takeaway

> **The Chi-Square test tells us whether the relationship is statistically detectable.**
>
> **Cramér's V tells us how strong that relationship is.**
>
> **Both are needed to understand the VTAT–cancellation relationship properly.**

### Next Step

We will calculate **Cramér's V** for the VTAT group × customer cancellation contingency table and interpret the magnitude of the effect.

````python
from scipy.stats import chi2_contingency
import numpy as np

# We already created contingency_table in 7A

chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

n = contingency_table.to_numpy().sum()

rows, cols = contingency_table.shape

cramers_v = np.sqrt(
    chi2 / (n * min(rows - 1, cols - 1))
)

print("Chi-square statistic:", chi2)
print("p-value:", p_value)
print("Sample size:", n)
print("Cramer's V:", cramers_v)
````

# Step 7C — New vs Repeat Customers 👥

Now we test **Hypothesis H3**.

Our question is:

> **Is customer cancellation associated with whether the booking comes from a new customer or a repeat customer?**

From our earlier descriptive analysis, we observed:

| Customer Type | Cancellation Rate |
| ------------- | ----------------: |
| **New**       |         **6.99%** |
| **Repeat**    |         **7.84%** |

At first glance, repeat customers appear to have a higher cancellation rate than new customers.

However, a difference in observed percentages does **not automatically mean that the relationship is statistically significant**.

We therefore need a statistical test.

---

## Hypotheses

We will use a **Chi-Square Test of Independence** to determine whether customer type and cancellation status are statistically associated.

### **H₀ — Null Hypothesis**

> **Customer type and customer cancellation are independent.**

In other words, whether a customer is **new or repeat** is not associated with whether they cancel their booking.

### **H₁ — Alternative Hypothesis**

> **Customer type and customer cancellation are associated.**

In other words, cancellation behavior differs systematically between **new and repeat customers**.

---

## Significance Level

We will use:

$$
\alpha = 0.05
$$

Therefore:

* **p-value < 0.05** → Reject H₀
* **p-value ≥ 0.05** → Fail to reject H₀

A statistically significant result would provide evidence that **customer type and cancellation are associated**.

However, statistical significance alone does not tell us how large or practically important the difference is.

---

## What We Are Testing

The Chi-Square test will compare the observed cancellation patterns across the two customer groups:

```text
Customer Type
      │
      ├── New
      │
      └── Repeat
             │
             ▼
     Customer Cancellation
             │
        ┌────┴────┐
        │         │
       Yes        No
```

The key question is:

> **Is the difference between the 6.99% and 7.84% cancellation rates larger than what we would reasonably expect from random variation alone?**

If the test is statistically significant, we will have evidence of an association between **customer type** and **cancellation behavior**.

If it is not statistically significant, the observed difference may be consistent with sampling variation.

---

### 🎯 Next Step

We will construct the **Customer Type × Cancellation** contingency table and perform the **Chi-Square Test of Independence**.

After that, we can determine whether the observed difference between **new and repeat customers** is statistically significant.
m

````python
from scipy.stats import chi2_contingency

# Create 2x2 contingency table
customer_table = pd.crosstab(
    df["customer_type"],
    df["is_customer_cancelled"]
)

print("Contingency Table:")
print(customer_table)

# Chi-square test
chi2, p_value, dof, expected = chi2_contingency(
    customer_table
)

print("\nChi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p_value)

print("\nExpected counts:")
print(expected)
````

# Now: Multivariable Logistic Regression

# Important Modeling Decision: The Model Population

Before fitting the model, we need to define exactly which bookings are included.

## Why We Cannot Use `Booking_Value`

`Booking_Value` is missing for customer-cancelled rides.

For the first model, we will therefore **not use `Booking_Value`**.

## Why We Exclude `No Driver Found` Bookings

There are **10,500 bookings** where no driver was found.

These bookings do not have an observed `Avg_VTAT`.

Since VTAT is our primary explanatory variable, these observations cannot be used in a model that requires VTAT.

Therefore, our first logistic regression will use:

> **Bookings where VTAT is observable.**

This defines our **model population** explicitly.

### Important distinction

This does **not** mean that `No Driver Found` bookings are unimportant.

It means that they cannot contribute to this particular VTAT-based model because the key predictor is unavailable for them.

We will document this limitation and can examine these bookings separately as part of the broader analysis.

````python
import statsmodels.formula.api as smf

# -------------------------------------------------------
# STEP 7D — Logistic Regression
# -------------------------------------------------------

# Keep only bookings where VTAT is available
model_df = df[df["Avg_VTAT"].notna()].copy()

print("Model sample size:", len(model_df))

# Logistic regression
model = smf.logit(
    formula="""
        is_customer_cancelled
        ~ Avg_VTAT
        + hour
        + C(customer_type)
        + C(Vehicle_Type)
    """,
    data=model_df
).fit()

print(model.summary())
````

For VTAT:

```python
import numpy as np

An odds ratio of approximately **1.39** means:

> **For each 1-unit increase in VTAT, the odds of customer cancellation are multiplied by approximately 1.39, holding hour, customer type, and vehicle type constant.**

$$
(1.39 - 1)\times100 \approx 39\%
$$

> **A 1-unit increase in VTAT is associated with approximately 39% higher odds of customer cancellation, after controlling for hour, customer type, and vehicle type.**

### ⚠️ Important distinction

This is a **39% increase in odds**, not necessarily a 39% increase in the probability of cancellation.

Odds and probability are different quantities.

Therefore, we should avoid saying:

> ❌ "Cancellation probability increases by 39%."

Instead, we should say:

> ✅ **"Cancellation odds are approximately 39% higher per 1-unit increase in VTAT, holding the other model variables constant."**

Q1 ──┐
Q2 ──┤
Q3 ──┤──► Cancellation
Q4 ──┘

Each group gets its own estimated effect.
```

The quartile model therefore makes **fewer assumptions about the shape of the relationship**.

## 7F.3 — Compare Continuous vs Quartile VTAT Models

The previous robustness check suggests that the relationship between **VTAT and customer cancellation may not be linear**.

### Continuous VTAT model

Our original logistic regression treated `Avg_VTAT` as a continuous variable:

> **OR ≈ 1.39 per additional minute**

This specification assumes that each additional minute of VTAT has the **same effect on the log-odds of cancellation**, regardless of whether the customer is experiencing a very short or very long wait.

### Quartile VTAT model

We then treated VTAT as four categorical groups:

| VTAT group | Odds vs Q1 |
| ---------- | ---------: |
| Q1         |  Reference |
| Q2         |  **8.49×** |
| Q3         |  **8.61×** |
| Q4         | **29.14×** |

All three quartile effects are highly statistically significant.

This suggests that the relationship may be better described as a **nonlinear or threshold-like pattern**:

> **Very short wait → low cancellation risk**
> **Moderate wait → large increase in cancellation odds**
> **Longest wait → very large increase in cancellation odds**

However, the descriptive pattern and individual coefficient significance are not enough by themselves to determine which model is more appropriate.

### Formal model comparison

We will compare:

* **Model A:** VTAT as a continuous variable
* **Model B:** VTAT as four categorical quartiles

Because Model B is more flexible, we want to determine whether that additional flexibility provides a meaningful improvement in model fit.

We can compare the models using a **likelihood-ratio test**.

### Hypotheses

**H₀:** The simpler continuous specification is sufficient; the additional quartile parameters do not significantly improve model fit.

**H₁:** The quartile specification provides a significantly better fit, suggesting that the effect of VTAT is not adequately represented by a single linear term.

### Interpretation

If the likelihood-ratio test is significant:

> The quartile model provides a statistically significant improvement in fit, supporting a nonlinear representation of the VTAT–cancellation relationship.

If the test is not significant:

> There is not enough statistical evidence that the additional flexibility of the quartile model improves fit, so the simpler continuous representation may be adequate.

This comparison allows us to choose the model based on **statistical evidence rather than visual inspection alone**.


````python
print("Continuous VTAT model")
print("AIC:", model_location.aic)
print("Log-Likelihood:", model_location.llf)

print("\nQuartile VTAT model")
print("AIC:", nonlinear_model.aic)
print("Log-Likelihood:", nonlinear_model.llf)
````

## 7F.4 — Model Selection and Next Step

The formal model comparison strongly favors the **continuous VTAT specification**.

| Model               |          AIC | Log-Likelihood |
| ------------------- | -----------: | -------------: |
| **Continuous VTAT** | **61,515.9** |  **−30,572.9** |
| Quartile VTAT       |     66,912.0 |      −33,269.0 |

Lower AIC indicates a better trade-off between model fit and model complexity.

The continuous model has an AIC approximately **5,396 points lower** than the quartile model, indicating that the additional flexibility of the quartile specification does **not** improve the overall model sufficiently.

### Decision

We will therefore keep **continuous `Avg_VTAT` as the main regression specification**.

The quartile model remains useful as a **descriptive and business-interpretation tool**, because it clearly illustrates that cancellation odds are much higher among customers experiencing longer waits.

However, we will **not use the quartile model as our final regression model** simply because its individual odds ratios are more dramatic.

### An important nuance

Choosing the continuous model does **not** prove that the relationship between VTAT and cancellation is perfectly linear.

The continuous logistic model assumes a constant effect of each additional minute **on the log-odds scale**. The quartile results suggest that the actual relationship may contain some curvature or a threshold effect.

Therefore, the next question is:

> **Can we capture possible nonlinearity while retaining VTAT as a continuous variable?**

This is preferable to relying on arbitrary quartile boundaries.

### Next robustness check

We will test a **flexible nonlinear specification** of continuous VTAT.

This will allow the relationship to bend rather than forcing one constant slope.

If the nonlinear specification provides a meaningful improvement in model fit, we will report that the VTAT–cancellation relationship is **nonlinear**.

If it does not improve the model materially, we can retain the simpler continuous specification with greater confidence.

This gives us a stronger modeling sequence:

**Continuous model → Quartile diagnostic → Formal model comparison → Flexible nonlinear test → Final specification**


### Step 7F.4 — Test a flexible VTAT relationship



````python
from patsy import bs
import statsmodels.formula.api as smf

# Flexible VTAT model
spline_model = smf.logit(
    formula="""
        is_customer_cancelled
        ~ bs(Avg_VTAT, df=4)
        + hour
        + C(customer_type)
        + C(Vehicle_Type)
        + C(Pickup_Location)
    """,
    data=model_df
).fit()

print("Continuous model AIC:", model_location.aic)
print("Spline model AIC:", spline_model.aic)

print("\nContinuous model log-likelihood:", model_location.llf)
print("Spline model log-likelihood:", spline_model.llf)
````

## 7F.5 — Convert the Spline Model into Business-Friendly Predictions

The spline model provides a substantially better fit than the simple continuous VTAT model:

| Model           |          AIC |
| --------------- | -----------: |
| Continuous VTAT |     61,515.9 |
| **Spline VTAT** | **54,054.6** |

The improvement of approximately **7,461 AIC points** provides strong evidence that the relationship between VTAT and customer cancellation is not adequately represented by a single straight-line effect.

### Why we should not interpret the spline coefficients directly

Spline models represent a nonlinear relationship using several basis functions.

As a result, the individual spline coefficients are difficult to interpret in business terms.

Instead of asking:

> "What does each spline coefficient mean?"

we ask:

> **"According to the model, how does the predicted probability of customer cancellation change as VTAT increases?"**

This gives us a much more intuitive business interpretation.

### Adjusted predicted probabilities

We will calculate predicted cancellation probabilities at representative VTAT values while holding the other model variables constant.

This allows us to translate the statistical relationship into statements such as:

> "At a short VTAT, the model estimates a relatively low cancellation probability."

and

> "At a substantially longer VTAT, the estimated cancellation probability increases considerably."

These are **model-adjusted predictions**, not raw cancellation rates.

### Important distinction

The predictions should not be interpreted as:

> "Every customer with this VTAT has exactly this probability of cancelling."

Instead, they represent the cancellation probability predicted by the fitted model under a specified set of conditions.

They are most useful for understanding the **shape and practical magnitude of the VTAT–cancellation relationship**.

### Business objective

The goal is to identify whether there is a range of VTAT where cancellation risk begins to increase sharply.

That can potentially translate the statistical finding into an operational question:

> **At what waiting-time range does reducing VTAT appear most valuable for protecting completed bookings?**


````python
# -------------------------------------------------------
# STEP 7F.5 — Predicted cancellation probability by VTAT
# -------------------------------------------------------

vtat_values = [2, 5, 8, 11, 14, 17, 20]

prediction_rows = []

for vtat in vtat_values:
    temp = model_df.copy()
    temp["Avg_VTAT"] = vtat
    
    predicted_probability = spline_model.predict(temp).mean()
    
    prediction_rows.append({
        "VTAT_minutes": vtat,
        "predicted_cancellation_probability": predicted_probability
    })

predicted_vtat = pd.DataFrame(prediction_rows)

predicted_vtat["predicted_cancellation_rate_pct"] = (
    predicted_vtat["predicted_cancellation_probability"] * 100
)

print(predicted_vtat)
````

Good catch. We do not accept that 97–100% result yet.

The spline may be capturing a very sharp pattern, but we need to compare its predictions against the actual cancellation rate at high VTAT values. Otherwise we risk reporting an overfit model.

## 7G — Data-Generating Rule and Model Validity Check

Before treating the regression results as genuine business evidence, we performed an additional sanity check on the relationship between VTAT and customer cancellation.

The results revealed an important structural pattern:

|   VTAT | Bookings | Customer cancellations | Cancellation rate |
| -----: | -------: | ---------------------: | ----------------: |
|     14 |    8,668 |                    771 |              8.9% |
|     15 |    3,847 |                    594 |             15.4% |
| **16** |  **821** |                **821** |          **100%** |
| **17** |  **627** |                **627** |          **100%** |
| **18** |  **768** |                **768** |          **100%** |
| **19** |  **626** |                **626** |          **100%** |
| **20** |  **403** |                **403** |          **100%** |

The key finding is that **every observed booking with VTAT from 16–20 minutes was cancelled by the customer**.

### Why this matters

A real-world marketplace relationship could certainly show increasing cancellation as waiting time increases.

However, a perfectly clean boundary where cancellation reaches **100% from exactly 16 minutes onward** is unusual and raises a data-generation concern.

It suggests that VTAT may be directly tied to the cancellation outcome through a deterministic rule or strong data construction process, for example:

> `VTAT >= 16 → customer cancellation`

We cannot prove the exact rule from this dataset alone, but the observed pattern is sufficiently strong that the regression results should **not automatically be interpreted as independently discovered marketplace behavior**.

### Impact on the statistical models

This finding changes how we interpret our earlier results.

The following analyses remain valuable as **statistical modeling practice**:

* Chi-square test
* Cramér's V
* Logistic regression
* Odds ratios
* Location-controlled regression
* Spline regression

However, their extremely strong VTAT results may largely reflect the underlying deterministic structure of the dataset.

In particular, the dramatic improvement from the spline model may not represent the discovery of a sophisticated nonlinear marketplace relationship.

Instead, the model may simply be learning the sharp boundary already embedded in the data.

### Analytical decision

We will therefore **not present the VTAT regression results as causal or independently validated business evidence**.

We will also avoid making claims such as:

> "Waiting 16+ minutes causes customers to cancel."

That conclusion is not justified by this dataset.

Instead, we will document the finding as a **data-quality and data-generating-process limitation**.

### Key lesson

This is an important lesson in practical data science:

> **A model can have excellent statistical fit and still produce misleading business conclusions if the underlying data contains artificial or deterministic relationships.**

Model performance must therefore be evaluated together with:

**data quality → data-generating process → assumptions → statistical evidence → business interpretation**

rather than relying on model fit alone.

## 7G.1 — Check for Separation

### What is separation?

In logistic regression, **separation** occurs when one or more predictors can almost perfectly distinguish the outcome classes.

In our data, we have already identified a particularly strong pattern:

> **VTAT ≥ 16 minutes → 100% customer cancellation**

This means that VTAT is capable of almost perfectly separating cancellations from non-cancellations in part of the dataset.

That can cause problems for logistic regression because the model may try to assign extremely large coefficients to reproduce the near-perfect boundary.

As a result:

* coefficients can become unstable or extremely large
* standard errors can behave unusually
* odds ratios can become misleading
* p-values and confidence intervals may not have their usual interpretation
* flexible models such as splines can exploit the boundary extremely effectively

Therefore, the very strong VTAT results we observed should be interpreted with caution.

### Practical diagnostic

We can first inspect cancellation rates at each VTAT value:

```python
# -------------------------------------------------------
# STEP 7G.1 — Check for Separation
# -------------------------------------------------------

    model_df
    .groupby("Avg_VTAT")["is_customer_cancelled"]
    .agg(
        bookings="count",
        cancellations="sum",
        cancellation_rate="mean"
    )
    .reset_index()
)

vtat_check["cancellation_rate_pct"] = (
    vtat_check["cancellation_rate"] * 100
)

print(vtat_check.to_string(index=False))
```

Now identify VTAT values where the outcome is perfectly determined:

```python

]

print("VTAT values with 0% or 100% cancellation:")

```

If the output confirms that VTAT values from approximately 16 minutes onward have a **100% cancellation rate**, we have direct evidence of outcome separation.

# -------------------------------------------------------

exposure = (
    model_df

exposure

> **Where cancellation volume is concentrated in the observed data.**

That is a defensible business-sizing metric even when causal interpretation is limited.


````python
# -------------------------------------------------------
# STEP 8A — Business Impact: High-VTAT Exposure
# -------------------------------------------------------

impact_df = df[df["Avg_VTAT"].notna()].copy()

impact_df["vtat_quartile"] = pd.qcut(
    impact_df["Avg_VTAT"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"],
    duplicates="drop"
)

impact_summary = (
    impact_df.groupby("vtat_quartile", observed=True)
    .agg(
        bookings=("row_id", "count"),
        customer_cancellations=("is_customer_cancelled", "sum"),
        cancellation_rate=("is_customer_cancelled", "mean")
    )
    .reset_index()
)

total_customer_cancellations = (
    impact_summary["customer_cancellations"].sum()
)

impact_summary["share_of_customer_cancellations_pct"] = (
    impact_summary["customer_cancellations"]
    / total_customer_cancellations
    * 100
)

print(impact_summary)
````

What we found

The highest-waiting 25% of bookings are:

34,106 bookings = 24.4% of all VTAT-observed bookings
They generated 6,046 customer cancellations
That's 57.6% of all customer cancellations

So:

Roughly one-quarter of bookings accounted for more than half of customer cancellations.

That is much more useful to a manager than simply saying “the p-value is tiny.

````python
# -------------------------------------------------------
# STEP 8B — Excess cancellation exposure
# -------------------------------------------------------

q1_rate = impact_summary.loc[
    impact_summary["vtat_quartile"] == "Q1",
    "cancellation_rate"
].iloc[0]

q4_bookings = impact_summary.loc[
    impact_summary["vtat_quartile"] == "Q4",
    "bookings"
].iloc[0]

q4_actual_cancellations = impact_summary.loc[
    impact_summary["vtat_quartile"] == "Q4",
    "customer_cancellations"
].iloc[0]

q4_expected_at_q1_rate = q4_bookings * q1_rate

excess_cancellations = (
    q4_actual_cancellations
    - q4_expected_at_q1_rate
)

print("Q1 benchmark cancellation rate:", q1_rate)
print("Q4 bookings:", q4_bookings)
print("Q4 actual cancellations:", q4_actual_cancellations)
print(
    "Q4 expected cancellations at Q1 rate:",
    q4_expected_at_q1_rate
)
print(
    "Excess cancellations vs Q1 benchmark:",
    excess_cancellations
)
````

What we found

Q4 had:

6,046 actual customer cancellations
If Q4 had the same cancellation rate as Q1, we'd expect about 251
Difference = ~5,795 excess cancellations relative to the Q1 benchmark

Again, this does not mean 5,795 cancellations are definitely avoidable. It is a benchmark comparison.

Because we've already identified synthetic behavior in this dataset, this number is for analytical demonstration only, not a real company's lost-booking estimate.

The largest non-completion category is **driver cancellation**, at 18% of all bookings.

### Where customer cancellation is concentrated

Although customer cancellation represents 7% of all bookings, it is not evenly distributed across VTAT groups.

The highest-VTAT quartile represents approximately:

* **24% of bookings**
* **57.6% of customer cancellations**

This is an important descriptive finding.

It means that the highest-waiting segment accounts for a **disproportionately large share of observed customer cancellation volume**.

This does not establish that reducing VTAT would necessarily prevent those cancellations, because the dataset contains a strong deterministic relationship between VTAT and cancellation.

However, it identifies **high-VTAT bookings as an important concentration point for cancellation exposure**.

### What the other analyses tell us

#### Customer type

New vs repeat customers did not show a statistically significant difference in cancellation behavior:

> **p = 0.275**

Therefore, customer type does not provide strong statistical evidence for explaining cancellation differences in this dataset.

#### Peak vs off-peak

Cancellation rates were essentially similar between peak and off-peak periods.

This suggests that time-of-day alone does not explain much of the observed cancellation behavior.

#### Booking value

Booking value cannot be reliably used to estimate the financial impact of customer cancellations because the variable is missing for cancelled bookings.

Therefore, we should **not estimate lost revenue from `Booking_Value`** using the observed values.

### Overall business interpretation

The dataset suggests that the broader operational problem is larger than customer cancellation alone.

The observed booking outcomes show substantial exposure across:

1. **Driver-side cancellations**
2. **Customer cancellations**
3. **No-driver-found bookings**
4. **Incomplete rides**

Within customer cancellations, the highest-VTAT segment represents a particularly large concentration of cancellation volume.

Therefore, the business story is not simply:

> "Customers cancel when they wait too long."

A more defensible interpretation is:

> **"The dataset shows substantial booking loss across both demand-side and supply-side failure categories. Customer cancellations account for 7% of bookings, while driver cancellations account for 18%. Within customer cancellations, the highest-VTAT quartile represents roughly 24% of bookings but 57.6% of customer cancellations, indicating a strong concentration of cancellation exposure among the longest-waiting bookings. However, the deterministic VTAT pattern limits causal interpretation."**

This keeps the analysis focused on **where the observed operational exposure is concentrated**, rather than claiming that the analysis has identified a causal solution.
